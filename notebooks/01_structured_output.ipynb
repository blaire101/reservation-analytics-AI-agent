## Goal

Convert a natural-language question into a small structured business object.

Production code: `app/core/models.py` and `app/core/extractor.py`.

```python
from app.settings import load_settings
from app.core.extractor import RequestExtractor

settings = load_settings('config/default.env')
extractor = RequestExtractor(settings)

question = 'How many users reserved Phone Mi 17 Pro in Germany for CMP001?'
request = extractor.extract(question)
request
```

The important idea is not the regex fallback. The important interface is:

`Natural Language → ExtractedRequest → ReservationQuery`

With `USE_LLM=true`, the same Pydantic model is filled by a structured LLM call.

```python
print(request.intent)
print(request.metric)
print(request.query.model_dump())
```
