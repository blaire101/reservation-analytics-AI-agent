## 3. IndexFlatL2

Flat = exact search over all stored vectors. L2 = Euclidean-distance ranking. It is simple and exact, which makes it a good learning choice for a small knowledge base.

## 4. Pure FAISS Example

````python
import numpy as np, faiss

vectors = np.array([[1,0,0],[0.9,0.1,0],[0,1,0]], dtype="float32")
query = np.array([[1,0.05,0]], dtype="float32")
index = faiss.IndexFlatL2(3)
index.add(vectors)
distances, indices = index.search(query, k=2)
print(distances, indices)
````

## 5. L2 vs Cosine Similarity

L2: smaller is closer. Inner product: larger is more similar. Cosine compares direction and is often implemented with normalized vectors plus inner product.

## Classic Architecture Q&A — Memorize This

### Q. Why do you use LangChain selectively instead of making it the entire application framework?

> **I use LangChain selectively rather than making it the entire application framework. LangChain's ChatOpenAI integration handles structured extraction, LangGraph handles stateful workflow orchestration, and LlamaIndex with FAISS handles the knowledge RAG layer.
